```python
model_path = "../../../../model/autophagy_september26/"
model_ginsim = ginsim.load(model_path + "Autophagy_and_apoptosis_Sept26.zginml")
ginsim.show(model_ginsim)
```

```python
## MaBoSS simulations
model_maboss = maboss.loadBNet(model_path + "Autophagy_and_apoptosis_Sept26.bnet")

# Set the default simulation parameters
model_maboss.update_parameters(sample_count=5000,max_time=50, time_tick=1)

# Define custom output phenotype colour mapping
phenotypes_colors = {
    'Apoptosis': 'green',
    'Proliferation': 'red',
    'BECN1': 'blue',
    'Autop_digestion': 'black'
}

# Define the color palette based on these colours
model_maboss.palette = phenotypes_colors

# Set the output nodes for the simulation
model_maboss.network.set_output(("Apoptosis", "Proliferation", "BECN1", "Autop_digestion"))
```

```python
# Set the baseline input values for the simulation
BASE_INPUTS = {"AA_Starvation": 0, "Growth_factor": 0, "TNF_R_or_DR": 0, "O_stress": 0,
               "Glucose_starv": 0, "Insuline": 0}

# Set the initial states for the simulation
for nodes in model_maboss.network.names:
    if nodes in BASE_INPUTS.keys():
        model_maboss.network.set_istate(nodes, [1 - BASE_INPUTS[nodes], BASE_INPUTS[nodes]])
    else:
        model_maboss.network.set_istate(nodes, [1,0])

# Run the simualtion
run_model_maboss = model_maboss.run()
run_model_maboss.plot_node_trajectory()
```
